# Altair on Avenger: interactive views
Each frame is drawn by Avenger in the kernel; the browser shows the PNG and sends the pointer back.

```python
import altair as alt, pandas as pd, numpy as np
import avenger_altair as av
av.enable()
df = pd.DataFrame({'category': list('ABCDEFGH'), 'amount': [3, 5, 2, 8, 4, 6, 7, 1]})
bars = alt.Chart(df).mark_bar().encode(x='category:N', y=alt.Y('sum(amount):Q', title='total')).properties(width=360, height=240)
```

```python
# The fisheye's focus follows the pointer.
av.view(bars.camera_fisheye(radius=0.4), scale=2.5, zoom=1.6)
```

```python
# Drag to turn the tilt.
av.view(bars.camera_tilt(), scale=2.5, zoom=1.6)
```

```python
# A histogram that grows while you look at it, through a fisheye.
rng = np.random.default_rng(0)
hist = alt.Chart(pd.DataFrame({'value': rng.normal(0, 1, 10_000)})).mark_bar().encode(alt.X('value:Q', bin=alt.Bin(maxbins=20)), y='count()').properties(width=420, height=240).camera_fisheye(radius=0.3)
live = av.live(hist)
live_view = live.view(scale=2.5, zoom=1.6)
live_view
```

```python
import time
for _ in range(100):
    live.append(pd.DataFrame({'value': rng.normal(0.5, 0.7, 10_000)}))
    time.sleep(0.05)
```
